```python
import numpy as np
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.kernel_ridge import KernelRidge
from sklearn.metrics import mean_squared_error


data_X, data_y = load_diabetes(return_X_y=True)
X_tr, X_te, y_tr, y_te = train_test_split(data_X, data_y, test_size=0.2, random_state=42)

# Part (a) — standard OLS linear regression
lin = LinearRegression()
lin.fit(X_tr, y_tr)
mse_lin = mean_squared_error(y_te, lin.predict(X_te))
print(f"Linear Regression MSE:      {mse_lin:.4f}")

# Q3(a) Result:
# Linear Regression MSE: 2900.19
# Linear regression finds the best fit hyperplane by minimising the sum
# of squared residuals. It assumes a purely linear relationship between
# the 10 features and the target which works reasonably well here.

# Part (b) — kernel ridge regression with two different kernels

# Kernel 1: Polynomial (degree=2)
# maps features into higher dimensional space to capture pairwise
# interactions between features that plain linear regression misses
krr_poly = KernelRidge(kernel='poly', degree=2, alpha=1.0)
krr_poly.fit(X_tr, y_tr)
mse_poly = mean_squared_error(y_te, krr_poly.predict(X_te))
print(f"KRR Polynomial MSE:         {mse_poly:.4f}")

# Kernel 2: RBF / Gaussian kernel
# measures similarity between points using a gaussian function and
# can model smooth nonlinear relationships in the data
# gamma=0.1 controls how quickly similarity drops off with distance
krr_rbf = KernelRidge(kernel='rbf', gamma=0.1, alpha=1.0)
krr_rbf.fit(X_tr, y_tr)
mse_rbf = mean_squared_error(y_te, krr_rbf.predict(X_te))
print(f"KRR RBF MSE:                {mse_rbf:.4f}")

# Q3(b) Comparison:
# Linear Regression MSE:      2900.19  <-- best
# KRR Polynomial (deg=2) MSE: 3969.78
# KRR RBF (gamma=0.1) MSE:    3975.65
#
# Linear regression outperforms both kernel methods on this dataset.
# This is because the diabetes dataset only has ~353 training samples
# across 10 features. Kernel methods have much higher model complexity
# and need more data to generalise well — with this few samples they
# overfit despite the alpha=1.0 regularisation.
# Both the polynomial and RBF kernels perform similarly at around MSE
# 3970, which suggests the problem is data scarcity rather than kernel
# choice. With a larger dataset the kernel methods would likely beat
# linear regression by capturing nonlinear feature interactions.
```
